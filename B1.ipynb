```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
titanic = sns.load_dataset('titanic')
df = pd.read_csv('Titanic-Dataset.csv')
```

```python
print("5 rows of the Titanic dataset:")
print(titanic.head())
```

```python
print("Dataset information:")
print(titanic.info())
print("Missing values:")
print(titanic.isnull().sum())
print(titanic.isnull().sum())
most_null = titanic.isnull().sum().idxmax()
```

```python
print("Average age of passengers:")
print(titanic.describe()['age']['mean'])
print(f"Most ticket value: {df['Fare'].max()}")
```

```python
# 4. How many male and female passengers are there?
gender_counts = df['Sex'].value_counts()
print("Count of male and female passengers:")
print(gender_counts)
```

```python
# 5. Which passenger class (1, 2, 3) has the most passengers? Count and percentage.
class_counts = df['Pclass'].value_counts()
class_percentages = df['Pclass'].value_counts(normalize=True) * 100

print("Passenger count by class:")
print(class_counts)
print("\nPassenger percentage by class (%):")
print(class_percentages)
```

```python
# 6. Filter all female passengers. What percentage of the total do they represent?
female_passengers = df[df['Sex'] == 'female']
female_percentage = (len(female_passengers) / len(df)) * 100

print(f"Number of female passengers: {len(female_passengers)}")
print(f"Percentage of female passengers: {female_percentage:.2f}%")
```

```python
# 7. Filter 1st class passengers. What is their average fare?
first_class = df[df['Pclass'] == 1]
avg_fare_first_class = first_class['Fare'].mean()

print(f"Average fare for first class passengers: {avg_fare_first_class:.2f}")
```

```python
# 8.1. How many passengers are under 18 years old?
under_18 = df[df['Age'] < 18]
print(f"Number of passengers under 18: {len(under_18)}")
```

```python
# 8.2. How many passengers are over 60 years old?
over_60 = df[df['Age'] > 60]
print(f"Number of passengers over 60: {len(over_60)}")
```

```python
import pandas as pd

# Giả lập dữ liệu doanh số cửa hàng coffee
df = pd.DataFrame({
    'CuaHang': ['Chi nhanh A', 'Chi nhanh B', 'Chi nhanh A', 'Chi nhanh C', 'Chi nhanh B'],
    'DoanhThu': [500, 300, 450, 600, 350],
    'Thang': [1, 1, 2, 1, 2]
})

# Tính tổng doanh thu theo từng cửa hàng
result = df.groupby('CuaHang')['DoanhThu'].sum().reset_index()
print(result)
```

```python
# Danh sách nhân viên và phòng ban
df_nv = pd.DataFrame({'ID': [1, 2, 3], 'Ten': ['An', 'Binh', 'Chi']})
df_pb = pd.DataFrame({'ID': [1, 2, 4], 'Phong': ['Sale', 'IT', 'HR']})

# Kết hợp để xem nhân viên thuộc phòng nào (Inner Join)
df_merge = pd.merge(df_nv, df_pb, on='ID', how='inner')
print(df_merge)
```

```python
# Dữ liệu bán máy tính điện ảnh
df_sales = pd.DataFrame({
    'KhuVuc': ['Mien Bac', 'Mien Bac', 'Mien Nam', 'Mien Nam'],
    'Loai': ['Laptop', 'Desktop', 'Laptop', 'Desktop'],
    'SoLuong': [10, 5, 15, 8]
})

# Tạo bảng so sánh số lượng bán theo khu vực và loại máy
pivot = df_sales.pivot_table(values='SoLuong', index='KhuVuc', columns='Loai', aggfunc='sum')
print(pivot)
```

```python
df_diem = pd.DataFrame({'Ten': ['Khoa', 'Mai', 'Lan'], 'Diem': [4, 8, 6]})

# Hàm phân loại học lực
def phan_loai(diem):
    return 'Dat' if diem >= 5 else 'Truot'

# Áp dụng hàm để tạo cột mới
df_diem['KetQua'] = df_diem['Diem'].apply(phan_loai)
print(df_diem)
```

```python
# Dữ liệu nhiệt độ các ngày trong tuần
df_weather = pd.DataFrame({
    'ThanhPho': ['Hanoi', 'HCM'],
    'Thu2': [30, 35],
    'Thu3': [32, 34]
})

# Chuyển các cột Thứ thành một cột "Ngay" duy nhất
df_long = df_weather.melt(id_vars='ThanhPho', var_name='Ngay', value_name='NhietDo')
print(df_long)
```

```python
# 4. How many male and female passengers are there?
gender_counts = df['Sex'].value_counts()
print("Count of male and female passengers:")
print(gender_counts)
```

```python
# 5. Which passenger class (1, 2, 3) has the most passengers? Count and percentage.
class_counts = df['Pclass'].value_counts()
class_percentages = df['Pclass'].value_counts(normalize=True) * 100

print("Passenger count by class:")
print(class_counts)
print("\nPassenger percentage by class (%):")
print(class_percentages)
```

```python
# 6. Filter all female passengers. What percentage of the total do they represent?
female_passengers = df[df['Sex'] == 'female']
female_percentage = (len(female_passengers) / len(df)) * 100

print(f"Number of female passengers: {len(female_passengers)}")
print(f"Percentage of female passengers: {female_percentage:.2f}%")
```

```python
# 7. Filter 1st class passengers. What is their average fare?
first_class = df[df['Pclass'] == 1]
avg_fare_first_class = first_class['Fare'].mean()

print(f"Average fare for first class passengers: {avg_fare_first_class:.2f}")
```

```python
# 8.1. How many passengers are under 18 years old?
under_18 = df[df['Age'] < 18]
print(f"Number of passengers under 18: {len(under_18)}")
```

```python
# 8.2. How many passengers are over 60 years old?
over_60 = df[df['Age'] > 60]
print(f"Number of passengers over 60: {len(over_60)}")
```

Tỉ lệ sống sót toàn tàu

```python
survival_rate = df['Survived'].mean() * 100
print(f"Tỉ lệ sống sót toàn tàu: {survival_rate:.2f}%")
```

Giải thích ý nghĩa: Tỉ lệ sống sót chỉ đạt khoảng hơn 38%, cho thấy thảm họa chìm tàu Titanic là một tai nạn nghiêm trọng với tỉ lệ tử vong cao. Số người thiệt mạng chiếm hơn 60% tổng số hành khách.

Dùng groupby: tỉ lệ sống sót theo giới tính và theo hạng vé

```python
survival_by_sex_class = df.groupby(['Sex', 'Pclass'])['Survived'].mean() * 100
print("Tỉ lệ sống sót theo giới tính và hạng vé (%):")
print(survival_by_sex_class)
```

Nhận xét so sánh:
- Theo giới tính: Tỉ lệ sống sót của phụ nữ cao vượt trội so với nam giới ở mọi hạng vé. Điều này phản ánh quy tắc "Phụ nữ và trẻ em lên trước" trong quá trình cứu nạn.
- Theo hạng vé: Có sự chênh lệch rõ ràng về tỉ lệ sống sót giữa các hạng vé. Hành khách hạng 1 (khoang hành khách cao cấp) có tỉ lệ sống sót cao nhất, tiếp đến là hạng 2 và thấp nhất là hạng 3. Điều này có thể do vị trí khoang của hạng 1 nằm gần boong tàu và xuồng cứu sinh hơn.
- Kết hợp cả hai yếu tố, phụ nữ hạng 1 có tỉ lệ sống sót gần như tuyệt đối, trong khi nam giới hạng 3 có tỉ lệ sống sót thấp nhất.

Trả lời Phần A - Câu 1: Trực quan số lượng hành khách nam và nữ

```python
gender_counts = df['Sex'].value_counts()
print("Số lượng hành khách theo giới tính:")
print(gender_counts)
```

Giải thích kết quả: Dựa vào bảng phân bố, số lượng khách nam gấp gần đôi số lượng khách nữ. Nam giới là lực lượng hành khách lớn nhất tham gia trên chuyến đi.

Trả lời Phần A - Câu 2: Số hành khách theo hạng vé (Pclass)

```python
class_counts = df['Pclass'].value_counts()
print("Số lượng hành khách ở mỗi hạng vé:")
print(class_counts)
```

Giải thích kết quả: Nhóm khách sử dụng hạng vé số 3 (hạng vé bình dân) chiếm số lượng áp đảo so với các phân khúc hành khách khác. Số khách của hạng 1 và hạng 2 cộng lại vẫn chưa bắt kịp số lượng hành khách hạng 3. Điều này cho thấy tầng lớp lao động chiếm tỷ trọng lớn trong chuyến đi này.

Trả lời Phần A - Câu 3: Thống kê số lượng phần trăm hành khách nữ

```python
female_passengers = df[df['Sex'] == 'female']
female_percentage = (len(female_passengers) / len(df)) * 100
print(f"Tổng số phần trăm hành khách nữ: {female_percentage:.2f}%")
```

Giải thích kết quả: Nữ giới chỉ chiếm hơn 35.24% số người tham gia chuyến hành trình trên số liệu báo cáo, điều đó có thể lý giải vì sao nhóm hành khách có tử suất cao chủ yếu là nằm ở nam giới. Mật độ nam giới đông hơn.

Đối chiếu dự đoán 1: Tỉ lệ người sống sót là khoảng 50%

```python
actual_survival_rate = df['Survived'].mean() * 100
print(f"Tỉ lệ sống sót thực tế: {actual_survival_rate:.2f}%")
```

Kết luận: Sai. Tỉ lệ sống sót thực tế chỉ đạt khoảng 38.38%, thấp hơn đáng kể so với dự đoán. Thực tế thảm họa khốc liệt hơn dự báo 50% ban đầu.

Đối chiếu dự đoán 2: Trên tàu có khoảng 40% nữ giới

```python
actual_female_percentage = (df['Sex'] == 'female').mean() * 100
print(f"Tỉ lệ nữ giới thực tế: {actual_female_percentage:.2f}%")
```

Kết luận: Chưa hoàn toàn chính xác. Tỉ lệ nữ giới thực tế trên tàu là khoảng 35.24%. Con số này gần sát với mức dự đoán 40%, nhưng thực tế lượng nam giới vẫn đông hơn một chút so với dự tính.

Đối chiếu dự đoán 3: Có khoảng 10% khách là hạng sang (hạng 1)

```python
actual_class1_percentage = (df['Pclass'] == 1).mean() * 100
print(f"Tỉ lệ khách hạng sang (hạng 1) thực tế: {actual_class1_percentage:.2f}%")
```

Kết luận: Sai. Tỉ lệ hành khách hạng sang (hạng 1) thực tế lên tới hơn 24.24%, cao hơn gấp đôi so với mức dự kiến 10%. Điều này thể hiện rằng có một lượng tương đối lớn giới thượng lưu và người giàu tham gia trên chuyến đi này.

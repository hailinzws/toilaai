1.1 Từ quan trọng nhất là data hay E(experience) do phải có nó thì máy tính mới có dữ liệu để học hỏi và làm theo ý mình đc
Học máy là làm cho máy tính có thể học được cách làm những thứ mình muốn

1.2 
Spam Filter :
- Experience : Các tệp email đc gắn nhãn spam hoặc từng bị report
- Task performance : Tỉ lệ % các email đc loại là rác tức tỉ lệ chính xác của hệ thống
Netflix : 
- Task : Đề xuất bộ phim giữ người dùng xem phim nhiều nhất có thể
- Experience : Lịch sử xem phim hành vi tìm kiếm của người dùng
- Task performance : Tỉ lệ người dùng click vào phim trên gợi ý và lượng thời gian người đó xem
Shoppe : 
- Task : Gợi ý các sản phẩm phù hợp với nhu cầu và sở thích mua sắm của từng khách hàng trên trang chủ/mục
- Experience : Lịch sử tìm kiếm sản phẩm, các mặt hàng đã xem, sản phẩm đã thêm vào giỏ hàng, và lịch sử các đơn hàng đã mua thành công trước đó.
- Task performance : Tỉ lệ chuyển đổi mua hàng từ danh sách gợi ý (Conversion Rate), hoặc doanh thu tăng trưởng nhờ vào mục gợi ý.

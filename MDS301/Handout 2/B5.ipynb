5.1
Training Set	
8,000 dòng (80%)	
Dùng để huấn luyện mô hình, giúp thuật toán học các quy luật, đặc trưng và trọng số từ dữ liệu.

Validation Set	
1,000 dòng (10%)	
Dùng để đánh giá mô hình trong quá trình thử nghiệm, tinh chỉnh siêu tham số (hyperparameters) và chọn ra mô hình tốt nhất.

Test Set	
1,000 dòng (10%)	
Dùng để kiểm tra và đánh giá khách quan hiệu năng cuối cùng của mô hình trên dữ liệu hoàn toàn mới trước khi triển khai thực tế.

Nếu bạn "nhìn" hoặc sử dụng tập Test để đưa ra các quyết định tinh chỉnh (tune) siêu tham số, bạn sẽ làm rò rỉ thông tin dữ liệu (Data Leakage). Lúc này, mô hình sẽ bị tối ưu hóa gián tiếp dựa trên tập Test. Kết quả đánh giá cuối cùng sẽ bị "ảo" (quá lạc quan so với thực tế), khiến bạn mất đi một thước đo khách quan và mô hình rất dễ bị overfitting (học vẹt), hoạt động kém khi gặp dữ liệu thực tế ngoài đời thực.

5.2 
Mô tả hoạt động theo các bước lặp chi tiết:

Chuẩn bị: Chia toàn bộ tập dữ liệu (thường là tập Train sau khi đã giữ riêng tập Test lớn) thành 5 phần (folds) bằng nhau và ngẫu nhiên, ký hiệu từ Fold 1 đến Fold 5.

Quá trình thực hiện lặp lại 5 lần (5 iterations):

Lần 1: Dùng Fold 1 làm tập Validation; các Fold 2, 3, 4, 5 gộp lại làm tập Training.
Lần 2: Dùng Fold 2 làm tập Validation; các Fold 1, 3, 4, 5 làm tập Training.
Lần 3: Dùng Fold 3 làm tập Validation; các Fold 1, 2, 4, 5 làm tập Training.
Lần 4: Dùng Fold 4 làm tập Validation; các Fold 1, 2, 3, 5 làm tập Training.
Lần 5: Dùng Fold 5 làm tập Validation; các Fold 1, 2, 3, 4 làm tập Training.

Kết quả cuối cùng của K-Fold: Là giá trị trung bình cộng (Average) của các điểm số hiệu năng (ví dụ như Accuracy, F1-score, hoặc Error) thu được từ cả 5 lần lặp ở trên.

Giảm thiểu tính ngẫu nhiên (Variance): Nếu chỉ dùng một tập validation cố định, có thể tập đó vô tình chứa toàn dữ liệu quá dễ hoặc quá khó, gây ra đánh giá sai lệch. K-Fold giúp loại bỏ sự may rủi này bằng cách cho mọi dữ liệu đều có cơ hội làm validation.

Tận dụng tối đa dữ liệu: Toàn bộ tập dữ liệu đều được luân phiên sử dụng để cả train lẫn validate. Điều này cực kỳ hữu ích khi lượng dữ liệu của bạn không quá lớn, giúp mô hình không bỏ sót các patterns quan trọng.

Đánh giá độ ổn định: Bằng cách nhìn vào độ lệch kết quả giữa các fold, bạn sẽ biết được mô hình của mình có chạy ổn định và bền vững khi phân phối dữ liệu thay đổi hay không.

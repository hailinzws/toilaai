3.1
Supervised:
Học từ dữ liệu có nhãn sẵn, tìm mối quan hệ giữa Input (X) và Output (Y) để dự đoán dữ liệu mới.
Có	
Lọc thư rác (Spam Filter), Dự đoán giá nhà, Nhận diện khuôn mặt.	
Linear Regression, SVM, KNN, Random Forest, Logistic Regression.

Unsupervised:
Không có teacher, tự tìm patterns, cấu trúc ẩn hoặc các cụm (clusters) tương đồng trong dữ liệu.	
Không	
Phân cụm khách hàng (Customer segmentation), Giảm số chiều dữ liệu.	
K-Means Clustering, PCA (Phân tích thành phần chính), Hierarchical Clustering.

Semi-supervised:
Kết hợp một lượng nhỏ dữ liệu có nhãn với một lượng lớn dữ liệu chưa có nhãn để tối ưu chi phí gán nhãn.	Một phần nhỏ có nhãn, phần lớn là không (Partially labeled)	
Phân loại ảnh trên Google Photos (nhận diện khuôn mặt qua vài ảnh có tên rồi tự nhóm các ảnh khác).	
Label Propagation, Self-Training, Co-training.

Reinforcement:
Mô hình (Agent) tự học bằng cách tương tác với môi trường qua cơ chế thử và sai (Trial and error) để tối đa hóa điểm thưởng.
Không (chỉ có reward)
Xe tự lái, AI chơi game (AlphaGo, Chess AI), Điều khiển cánh tay Robot.	Q-Learning, Deep Q-Network (DQN), PPO, SARSA.

3.2
Dự đoán giá nhà (triệu đồng)
R
Output là một giá trị số liên tục đại diện cho số tiền (Ví dụ: 2500 triệu, 3150.5 triệu...).

Email: spam hay không?
C
Output là nhãn phân loại nhị phân rời rạc (Chỉ có 2 lớp: Spam hoặc Không Spam).

Dự đoán điểm thi của sinh viên (0-10)
R
Output là một giá trị số liên tục nằm trong thang điểm (Ví dụ: 7.5, 8.25, 9.0...).

Chẩn đoán bệnh: ung thư / bình thường / lành tính
C
Output là nhãn phân loại đa lớp rời rạc (Chọn 1 trong 3 danh mục bệnh cụ thể).
Dự đoán số lượng hàng bán ngày mai
R
Output là một giá trị số đại diện cho số lượng sản phẩm tiêu thụ (Giá trị số có tính chất liên tục).

Nhận diện chữ viết tay: chữ số 0-9
C
Output là nhãn phân loại rời rạc gồm 10 lớp tương ứng với các chữ số cụ thể từ 0 đến 9.

3.3
Robot học chơi cờ vua — thắng/thua là reward:
Reinforcement
Robot đóng vai trò là Agent tự đưa ra quyết định (nước đi) và học từ kết quả phản hồi (thắng/thua) của môi trường dưới dạng điểm thưởng (reward).

Phân nhóm 10 triệu ảnh sản phẩm không có nhãn:
Unsupervised
Tập dữ liệu hoàn toàn không có nhãn định hướng trước. Mục tiêu là phân cụm (clustering) các bức ảnh dựa trên các đặc trưng tương đồng tự thân của chúng.

Dự đoán SV nào có nguy cơ bỏ học dựa trên điểm số lịch sử:
Supervised
Dữ liệu lịch sử đã có sẵn nhãn rõ ràng (Y: Bỏ học hoặc Học tiếp) gắn với các biến tính năng (X: Điểm số). Mô hình dựa vào đó để phân loại học sinh mới.

Nhận diện tiếng nói: có 1000 giờ audio có label + 100,000 giờ không label:
Semi-supervised
Tập dữ liệu có sự kết hợp giữa một lượng nhỏ dữ liệu đã được gán nhãn (1,000 giờ) và một lượng rất lớn dữ liệu chưa gán nhãn nhằm tối ưu chi phí.

Hệ thống gợi ý sản phẩm dựa trên lịch sử mua hàng:
Unsupervised
Thường sử dụng các thuật toán phân cụm hoặc lọc cộng tác (Collaborative Filtering) để tìm ra các nhóm khách hàng hoặc sản phẩm có chung mô thức hành vi.

Phát hiện bất thường trong dữ liệu server (log files):
Unsupervised
Đây là bài toán Anomaly Detection. Máy sẽ tự tìm hiểu cấu trúc của các log bình thường, từ đó tự động phát hiện và chỉ ra các mẫu dữ liệu dị biệt mà không cần nhãn lỗi trước.

Dịch tự động anh-việt với dataset 1M cặp câu:AlphaGo học cách đánh cờ vây:
Supervised
Tập dữ liệu chứa các cặp câu đối ứng rõ ràng: câu tiếng Anh đóng vai trò là Input (X) và câu dịch tiếng Việt tương ứng chính là nhãn mục tiêu (Y).

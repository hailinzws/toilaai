Overfitting
(Mô hình học vẹt dữ liệu train, mất khả năng tổng quát hóa)
- Áp dụng Regularization (L1/L2).
- Thêm tầng Dropout (nếu dùng Deep Learning).
- Giảm bớt số lượng feature nhiễu.
- Thu thập thêm dữ liệu huấn luyện hoặc dùng Early Stopping.

Underfitting
(Mô hình quá đơn giản, chưa học được quy luật của dữ liệu)
- Tăng độ phức tạp của mô hình (chọn thuật toán mạnh hơn, thêm layer/neuron).
- Giảm bớt các ràng buộc Regularization.
- Thực hiện Feature Engineering (tạo thêm biến mới từ dữ liệu gốc).

Good Fit
(Mô hình chạy ổn định, độ chính xác ở mức khá và chấp nhận được)
- Mô hình đạt trạng thái tốt. Có thể giữ nguyên và thử nghiệm tinh chỉnh nhẹ siêu tham số (Hyperparameter Tuning) để tối ưu thêm một chút nếu cần.

Underfitting
(Mô hình tuy không bị lệch giữa hai tập nhưng độ chính xác nền quá thấp)
- Đổi sang thuật toán mạnh hơn (ví dụ: từ Tuyến tính sang Tree-based như Random Forest/XGBoost).
- Thêm các tính năng (features) có độ tương quan cao hơn với biến mục tiêu.

Good Fit / Lý tưởng
(Mô hình xuất sắc, độ chính xác cao và tổng quát hóa cực kỳ tốt)
- Hoàn hảo. Lưu lại trọng số (weights/checkpoint) tốt nhất này và sẵn sàng đem mô hình đi triển khai thực tế (Deployment).

Depth

Depth tăng thì train tăng nhưng test giảm 

Depth gây overfitting rõ nhất: depth=10 và depth=None
Dấu hiệu nhận biết: 1.  Độ chính xác trên tập Train đạt tuyệt đối: Giá trị train tăng lên mức tối đa là 1.000 (100%). Điều này chứng tỏ cây quyết định đã chia quá nhiều nhánh nhỏ để "học vẹt" và khớp hoàn toàn với từng điểm dữ liệu huấn luyện (bao gồm cả nhiễu).

Tôi sẽ chọn depth 1 để deploy vì nó cho ra kết quả gần với train nhất tức là đạt đúng yêu cầu với train 

```python
import numpy as np
from sklearn.tree import DecisionTreeClassifier
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split

# Tạo dataset đơn giản
X, y = make_classification(n_samples=200, n_features=10, random_state=42)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3)

# Thử các độ sâu khác nhau
for depth in [1, 2, 3, 5, 10, None]:  # None = không giới hạn
    model = DecisionTreeClassifier(max_depth=depth, random_state=42)
    model.fit(X_train, y_train)
    train_acc = model.score(X_train, y_train)
    test_acc = model.score(X_test, y_test)
    print(f'depth={depth}: train={train_acc:.3f}, test={test_acc:.3f}')
```

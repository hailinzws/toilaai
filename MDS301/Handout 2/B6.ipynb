6.1
Bài toán tôi muốn giải quyết bằng ML:

Dự đoán xu hướng biến động giá (Lên / Xuống / Đi ngang) của một mã cổ phiếu cụ thể trong phiên giao dịch kế tiếp dựa trên dữ liệu lịch sử giá và khối lượng giao dịch.

Tại sao bài toán này quan trọng? (impact là gì?):

Giúp tối ưu hóa quyết định đầu tư, giảm thiểu rủi ro thua lỗ cho danh mục và tự động hóa tần suất giao dịch (Algorithmic Trading). Impact cụ thể là tăng tỷ lệ lợi nhuận danh mục (ROI) cao hơn mức trung bình của thị trường.

6.2 
Task (T) — Bài toán cụ thể:

Dự án tôi: Phân loại xu hướng giá cổ phiếu phiên ngày mai (T+1) thuộc nhóm Tăng (1) hay Giảm (0) dựa trên chuỗi thời gian (Time-series Classification).

Experience (E) — Data sẽ dùng:

Dự án tôi: Tập dữ liệu gồm bản ghi lịch sử giao dịch (bao gồm: Giá mở cửa, Giá đóng cửa, Giá cao nhất/thấp nhất, Khối lượng giao dịch - Volume, và các chỉ báo kỹ thuật như MA, RSI, MACD được tính toán từ dataset).

Performance (P) — Đo lường thành công:

Dự án tôi: Độ chính xác phân loại xu hướng Directional Accuracy (hoặc Precision cho lệnh Mua) đạt > 65% trên tập dữ liệu thử nghiệm (Test set), đồng thời thuật toán kiểm thử lịch sử (Backtesting) mang lại lợi nhuận dương. (Lưu ý: Trong tài chính, dự đoán giá cổ phiếu đạt >65% đã là mức xuất sắc do thị trường có độ nhiễu rất cao).

6.2
Loại learning nào phù hợp?

Supervised (Học có giám sát) — Lý do: Chúng ta có dữ liệu lịch sử trong quá khứ đã biết rõ nhãn (giá ngày hôm sau thực tế là tăng hay giảm) để làm mục tiêu (Target variable) cho mô hình học.

Classification hay Regression?

Classification (Nếu mục tiêu là dự đoán hướng đi: Tăng/Giảm để ra quyết định Mua/Bán).
(Hoặc có thể chọn Regression nếu mục tiêu của bạn là dự đoán chính xác con số giá đóng cửa là bao nhiêu nghìn đồng).

Data lấy từ đâu?

API (Ví dụ: Từ các thư viện tài chính như Yahoo Finance yfinance, hoặc API của các công ty chứng khoán Việt Nam như SSI, VnDirect) / Kaggle.

Thách thức lớn nhất bạn dự đoán là gì?

Data quality / Thừa nhiễu (Market Noise) / Hiện tượng quá khớp (Overfitting) do dữ liệu tài chính biến động liên tục và chịu ảnh hưởng lớn bởi tin tức vĩ mô không có trong số liệu.

Nếu model đạt 80% accuracy, bạn có hài lòng không?

Cực kỳ hài lòng nhưng sẽ rất nghi ngờ. Phụ thuộc vào: Hiện tượng rò rỉ dữ liệu (Data Leakage) khi chia tập Train/Test trên dữ liệu chuỗi thời gian (Time-series). Trong chứng khoán, nếu không phân tách theo thời gian một cách nghiêm ngặt, mô hình rất dễ "học trước tương lai" dẫn đến accuracy ảo lúc train nhưng lỗ nặng khi chạy thực tế.

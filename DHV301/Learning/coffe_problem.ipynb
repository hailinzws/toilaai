```python
import pandas as pd
import numpy as np
```

```python
df =pd.read_csv('dirty_cafe_sales.csv')
```

```python
df.info()
```

```python
missing_values = df.isnull().sum()
print(missing_values)
```

```python
duplicate_values = df.duplicated().sum()
print(f"Number of duplicate rows: {duplicate_values}")
```

```python
print(df.describe(include='all'))
```

```python
df['Transaction Date'] = df['Transaction Date'].fillna('UNKNOWN')

df['Quantity'] = pd.to_numeric(df['Quantity'], errors='coerce').astype('Int64')
df['Total Spent'] = pd.to_numeric(df['Total Spent'], errors='coerce').astype('float')
df['Price Per Unit'] = pd.to_numeric(df['Price Per Unit'], errors='coerce').astype('float')
df['Transaction Date'] = pd.to_datetime(df['Transaction Date'], errors='coerce')
```

```python
print(df['Item'].unique())
print(df['Payment Method'].unique())
print(df['Location'].unique())

print(df['Price Per Unit'].value_counts(dropna=False))
print(df['Quantity'].value_counts(dropna=False))
print(df['Total Spent'].value_counts(dropna=False))
```

```python
df_clean = df.copy()
error_word = ['UNKNOWN', 'nan', 'ERROR', ]
columns_to_replace = ['Item', 'Payment Method', 'Location']

for column in columns_to_replace:
    df_clean[column] = df_clean[column].replace(error_word, np.nan)

df_clean['Item'] = df_clean['Item'].fillna(df_clean['Item'].mode()[0])
df_clean['Location'] = df_clean['Location'].fillna(df_clean['Location'].mode()[0])
df_clean['Payment Method'] = df_clean['Payment Method'].fillna(df_clean['Payment Method'].mode()[0])   

df_clean['Quantity'] = df_clean['Quantity'].fillna(df_clean['Quantity'].median())
df_clean['Quantity'] = pd.to_numeric(df_clean['Quantity'], errors='coerce').round().astype('int64')

df_clean['Total Spent'] = df_clean['Total Spent'].fillna(df_clean['Total Spent'].median())
df_clean['Total Spent'] = pd.to_numeric(df_clean['Total Spent'], errors='coerce').round(2)

df_clean['Price Per Unit'] = df_clean['Price Per Unit'].fillna(df_clean['Price Per Unit'].median())
df_clean['Price Per Unit'] = pd.to_numeric(df_clean['Price Per Unit'], errors='coerce').round(2)

df_clean['Transaction Date'] = df_clean['Transaction Date'].fillna(df_clean['Transaction Date'].mode()[0])
df_clean['Transaction Date'] = pd.to_datetime(df_clean['Transaction Date'], errors='coerce')
```

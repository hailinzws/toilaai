```python
import pandas as pd
df= pd.read_csv("survey_results_public.csv")
df2=pd.read_csv("salary_survey_raw.csv")

print(f'Shape: {df.shape}')
```

```python
null_counts = df2.isnull().sum()
print("num total: ",null_counts)

cols_with_null = null_counts[null_counts > 0]

print("Các cột bị thiếu dữ liệu:")
print(cols_with_null)
```

```python
miss = df.isnull().sum()
pct  = (miss / len(df) * 100).round(1)
report = pd.DataFrame({'count': miss, 'pct_%': pct})
report = report[report['count'] > 0].sort_values('pct_%', ascending=False)
print(report)
```
